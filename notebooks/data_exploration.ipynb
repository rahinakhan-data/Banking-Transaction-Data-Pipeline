#### Task 4 – Load All Three Datasets

```python
# Import core utilities for file path handling and data manipulation
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Configure relative paths for source (raw_data) and target (processed) directories
raw_data_dir = '../raw_data/'
processed_dir = '../processed/'
graph_dir = '../reports/graphs/'
```

```python
# set the path
# raw_data_dir = "../raw_data/"
try:
    #1. Load all three CSV files
    print("Loading All Regional Datasets...")

    df_west = pd.read_csv(os.path.join(raw_data_dir, "west_transactions.csv"))
    df_north = pd.read_csv(os.path.join(raw_data_dir, "north_transactions.csv"))
    df_south = pd.read_csv(os.path.join(raw_data_dir, "south_transactions.csv"))

    print("SUCCESS. All regional datasets are loaded successfully.\n")

    datasets = {'West Region':df_west, 'North Region':df_north, 'South Region': df_south}

    for region_name, df in datasets.items():
        print("*"*60, region_name, "*"*60)

        # top 10 data records
        print("First 10 rows")
        display(df.head(10))

        # dataset dimensions (Shape)
        print("\nTotal rows & columns ",df.shape)

        # schema column headers
        print("\nDataset Columns:\n", list(df.columns))

        # column data types
        print("\nData types of each column\n",df.dtypes,"\n\n")

except FileNotFoundError as err:
    print("Error: File was not found. Check raw_data folder.\n", err)
    
except Exception as e:
    print("Unexpected Error occurred while reading the files: ", e)
```

#### Task 5 – Merge the Datasets

```python
# Merge all three regional datasets into one master DataFrame.
master_df = pd.concat([df_west, df_north, df_south], ignore_index=True)
print("\nSUCCESS: West, North, and South datasets are merged successfully!")
print(f"Master Dataset Shape: \nRows: {master_df.shape[0]}, Columns: {master_df.shape[1]}")

# Calculate the sum of rows from all three raw input datasets
total_raw_rows = len(df_west) + len(df_north) + len(df_south)

# Validate that the master dataset row count perfectly matches the total input count
if master_df.shape[0] == total_raw_rows:
    print("Verification Successful: All records merged successfully!")
else:
    print("Warning: Row counts do not match after merge!")
```

```python
# processed_dir = "../processed/"

# Create the complete file path by joining the folder name and file name
master_path = os.path.join(processed_dir, 'master_transactions.csv')

# Save the combined data as a CSV file
master_df.to_csv(master_path, index=False)

print("--- Exporting Master Dataset ---")
print(f"Master dataset successfully saved at: {master_path}")
```

#### Task 8 – Exploratory Data Analysis (EDA)

```python
# Path to save generated charts and figures
graph_dir = "../reports/graphs/"
```

```python
# Chart 1:  Transaction Count by Region
fig = plt.figure(figsize=(8,5))
sns.countplot(data=master_df, x='region', hue='region', palette='Set2')
plt.title("Transaction Count by Region")
plt.xlabel("Region")
plt.ylabel("Transaction Count")
plt.savefig(os.path.join(graph_dir, 'chart1_regional_transaction_volumes.png'), dpi=300)
```

```python
# Chart 2: Transaction Type Distribution
fig = plt.figure(figsize=(8,5))
sns.countplot(data=master_df, x='transaction_type', hue='transaction_type', palette='Set2')
plt.title("Transaction Type Distribution")
plt.xlabel("Transaction Type")
plt.ylabel("Total Count")
plt.savefig(os.path.join(graph_dir, 'chart2_transaction_type.png'), dpi=300)
```

```python
# Chart 3: Channel Distribution
fig = plt.figure(figsize=(8,5))
sns.countplot(data=master_df, x='channel', hue='channel', palette='Set2')
plt.title("Channel Disctribution")
plt.xlabel("Channel")
plt.ylabel("Total Counts")
plt.savefig(os.path.join(graph_dir, 'chart3_channel_distribution.png'), dpi=300)
plt.show()
```

```python
# Chart 4: Success vs Failed Transactions
fig = plt.figure(figsize=(8,5))
txn_status = master_df['status'].value_counts()
plt.pie(txn_status.values, labels = txn_status.index, autopct='%1.1f%%', startangle = 90, colors = ['#66c2a5', '#8da0cb', ])
plt.title("Transaction Success vs. Failure Rate")
plt.savefig(os.path.join(graph_dir, 'chart4_transaction_success_rate.png'), dpi =300)
```

```python
# Chart 5: Top 5 Branches by Number of Transactions
fig = plt.figure(figsize=(8,5))
top_branches_by_txn = master_df.groupby('branch_code')['transaction_id'].count().reset_index()
top_5_branches = top_branches_by_txn.sort_values(by='transaction_id', ascending=False).head(5)

plt.bar(x = top_5_branches['branch_code'], height = top_5_branches['transaction_id'], color = ['#66c2a5','#fc8d62', '#8da0cb','#e78ac3','#ffd92f' ])
plt.title("Top 5 Branches by Number of Transactions")
plt.xlabel("Branches")
plt.ylabel("Total Transaction counts")

plt.savefig(os.path.join(graph_dir, 'chart5_top_performing_branches.png'), dpi = 300)
plt.show()
```

```python
# Chart 6: Transaction Amount Distribution
fig = plt.figure(figsize=(8,5))
sns.histplot(data=master_df, x='amount', kde=True)
plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Total Counts")
plt.savefig(os.path.join(graph_dir, 'chart6_transaction_amount.png'), dpi = 300)
plt.show()
```
